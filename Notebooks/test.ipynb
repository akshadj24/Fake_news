```python
## load model and the tokinizer
from tensorflow.keras.models import load_model
import pickle
import nltk
from nltk.stem import WordNetLemmatizer

model=load_model("../model/m2.h5")

with open('tokenizer.pkl','rb') as file:
    t_N=pickle.load(file)
```

```python
import re 
from nltk.stem import WordNetLemmatizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

l1=WordNetLemmatizer()
```

```python
# Raw text preprocessor function 

def preprocessor(input):
    words=re.sub('[^a-zA-z]',' ',input.lower())
    words=words.split()
    word=[l1.lemmatize(j) for j in words]
    words=" ".join(word)

    #we pass to tokinzer
    seq=t_N.texts_to_sequences([words])
    pad_seq=pad_sequences(seq,maxlen=800)
    return pad_seq
```

```python
# model prediction

prob=model.predict(preprocessor('The government on Tuesday announced a new policy aimed at improving transparency in political funding. According to officials from the Ministry of Finance, all political parties will now be required to disclose details of donations received above a specified limit. The move is expected to strengthen democratic processes and reduce the influence of unaccounted money in elections. Opposition leaders welcomed the step but urged the government to ensure strict implementation and independent monitoring to prevent misuse'))
if prob >= 0.5:
        label = "Real"
else:
        label = "Fake"

print(F"The news is:{label}")
print(prob)
```
